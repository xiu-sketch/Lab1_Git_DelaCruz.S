```python
target_name = "DELA CRUZ"      
STUDENT_ID = "TUPM-25-0164"      
SEED_NUM = STUDENT_ID[-1]      
redacted_name = ""

vowels = "AEIOU"

# Redaction loop
for char in target_name:
    if char in vowels:
        redacted_name += SEED_NUM
    else:
        redacted_name += char

print("Original:", target_name)
print("SEED_NUM:", SEED_NUM)
print("Redacted:", redacted_name)
```

```python
# SAMPLE DATA
engineer_name = "SAMUEL"
first_letter = engineer_name[0]

A = ord(first_letter)     
B = int(SEED_NUM)         
C = len(target_name)      

verification_hash = (A * B) + C

print("First Letter ASCII:", A)
print("SEED_NUM:", B)
print("Total Loops:", C)
print("Verification Hash:", verification_hash)
```
